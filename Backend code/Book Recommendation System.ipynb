**Popularity Based Recommendation System**

```python
ratings_with_name = ratings.merge(books, on='ISBN')
```

```python
ratings_with_name.head()
```

```python
ratings_with_name.shape
```

```python
num_rating_df = ratings_with_name.groupby("Book-Title").count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating':'num_ratings'}, inplace=True)
num_rating_df
```

```python
avg_rating_df = ratings_with_name.groupby("Book-Title")['Book-Rating'].mean().reset_index()
avg_rating_df.rename(columns={'Book-Rating':'avg_ratings'}, inplace=True)
avg_rating_df
```

```python
popular_df = num_rating_df.merge(avg_rating_df , on="Book-Title")
```

```python
popular_df
```

```python
popular_df=popular_df[popular_df["num_ratings"]>=300].sort_values('avg_ratings',ascending=False).head(50)
```

```python
popular_df
```

```python
popular_df=popular_df.merge(books, on="Book-Title").drop_duplicates("Book-Title")[["Book-Title","Book-Author","Image-URL-M","num_ratings","avg_ratings"]]
```

```python
popular_df
```

**Collaborative Filtering Based Recommender System**

```python
ratings_with_name.groupby("User-ID").count()["Book-Rating"]
```

```python
ratings_with_name.groupby("User-ID").count()["Book-Rating"]>100
```

```python
x=ratings_with_name.groupby("User-ID").count()["Book-Rating"]>100
x[x]
```

```python
users100=x[x].index
```

```python
filtered_rating=ratings_with_name[ratings_with_name["User-ID"].isin(users100)]
```

```python
filtered_rating
```

```python
y=filtered_rating.groupby("Book-Title").count()["Book-Rating"]>=50
```

```python
famous_books = y[y].index
```

```python
famous_books
```

```python
filtered_rating[filtered_rating["Book-Title"].isin(famous_books)]
```

```python
final_ratings = filtered_rating[filtered_rating["Book-Title"].isin(famous_books)]
```

```python
final_ratings
```

```python
pt = final_ratings.pivot_table(index="Book-Title", columns="User-ID", values="Book-Rating")
```

```python
pt
```

```python
pt.fillna(0, inplace=True)
```

```python
pt
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_scores = cosine_similarity(pt)
```

```python
similarity_scores[0]
```

```python
np.where(pt.index=='1984')[0][0]
```

```python
sorted(list(enumerate(similarity_scores[0])), key=lambda x:x[1], reverse=True)[1:6]
```

```python
def recommend(book_name):
    matches = np.where(pt.index == book_name)[0]
    if len(matches) == 0:
        return []
    index = matches[0]
    similar_items = sorted(list(enumerate(similarity_scores[index])),
                           key=lambda x: x[1], reverse=True)[1:6]
    data = []
    for i in similar_items:
        temp_df = books[books['Book-Title'] == pt.index[i[0]]]
        title = temp_df['Book-Title'].values[0]
        author = temp_df['Book-Author'].values[0]
        image = temp_df['Image-URL-M'].values[0]
        data.append([title, author, image])
    return data
```

```python
recommend("1984")
```

```python
recommend("Harry Potter and the Prisoner of Azkaban (Book 3)")
```

```python
recommend("Message in a Bottle")
```

```python
recommend("The Da Vinci Code")
```

```python
recommend("You Belong To Me")
```

```python
recommend("Harry Potter and the Goblet of Fire (Book 4)")
```
